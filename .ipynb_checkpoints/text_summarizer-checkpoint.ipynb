```python
import pandas as pd
from transformers import T5Tokenizer, Trainer, TrainingArguments, T5ForConditionalGeneration
```

```python
train_data = pd.read_csv("samsum-train.csv")
val_data = pd.read_csv("samsum-validation.csv")
```

```python
# Random sampling
train_data = train_data.sample(n=4000, random_state=42).reset_index(drop=True) 
val_data = val_data.sample(n=500, random_state=42).reset_index(drop=True) 
```

```python
train_data.shape
```

## Data pre-processing

```python
import re
def clean_data(text):
    text = re.sub(r"\r\n", " ", text) # remove lines
    text = re.sub(r"\s+", " ", text)  # remove spaces
    text = re.sub(r"<.*?>", " ", text)  # remove html tags
    text = text.strip().lower()
    return text
```

```python
train_data["dialogue"] = train_data["dialogue"].apply(clean_data)
train_data["summary"] = train_data["summary"].apply(clean_data)

val_data["dialogue"] = val_data["dialogue"].apply(clean_data)
val_data["summary"] = val_data["summary"].apply(clean_data)
```

## Tokenize

```python
tokenizer = T5Tokenizer.from_pretrained("t5-small")
```

```python
def tokenize(data):
    inputs = tokenizer(
        data["dialogue"],
        padding="max_length",
        max_length=512,
        truncation=True
    )

    targets = tokenizer(
        data["summary"],
        padding="max_length",
        max_length=150,
        truncation=True
    )

    inputs["labels"] = targets["input_ids"]

    return inputs
```

```python
train_dataset = train_data.apply(tokenize, axis=1).tolist()
val_dataset = val_data.apply(tokenize, axis=1).tolist()
```

```python
train_dataset[0]
# input ids - dialogue => token ids
# 1 -->> EOS, 0s -->> Padding
# attention mask telling which are valid and which are padding values in input values
# labels - target => summary token
```

# Working with our model

```python
#NLP => generation task

model = T5ForConditionalGeneration.from_pretrained("t5-small")
```

```python
import torch
if torch.backends.mps.is_available():
    device = torch.device("mps")
elif torch.cuda.is_available():
    device = torch.device("cuda")
else:
    device = torch.device("cpu")
print("device: ", device)
model.to(device)
```

```python
# Training Arguements
training_args = TrainingArguments(
    output_dir = "./results",
    
    num_train_epochs=1,
    weight_decay=0.01,
    
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,

    eval_strategy="epoch",
    save_strategy="epoch",

    warmup_steps=100
    #0 => lr default
)
```

```python
# Trainer=> provides an api for featiue-complete training in Pytorch
trainer = Trainer(
    model=model,args=training_args,
    train_dataset=train_dataset,
    eval_dataset=val_dataset
)
```

```python
# Train the model
trainer.train()
```

```python
# model load => fine-tune => save the model
model.save_pretrained("./saved_summary_model")
tokenizer.save_pretrained("./saved_summary_model")
```

```python
model = T5ForConditionalGeneration.from_pretrained("./saved_summary_model")
tokenizer = T5Tokenizer.from_pretrained("./saved_summary_model")
```

## test for the core logic for summarization

```python
def summarize_dialogue(dialogue):
    dialogue = clean_data(dialogue)

    # Tokenize
    inputs = tokenizer(
        dialogue,
        padding="max_length",
        max_length=512,
        truncation=True,
        return_tensors="pt"
    ).to(device)

    # Generate summary
    model.to(device)

    targets = model.generate(
        input_ids=inputs["input_ids"],          # ✅ FIXED
        attention_mask=inputs["attention_mask"],
        max_length=150,
        num_beams=4,
        early_stopping=True
    )

    # Decode token IDs into text
    summary = tokenizer.decode(
        targets[0],
        skip_special_tokens=True
    )

    return summary
```

```python
test_dialogue = """
Reporter: In today's technology news, artificial intelligence continues to expand rapidly across industries, from healthcare to finance.

Reporter: Companies are investing heavily in machine learning systems to automate tasks, improve decision-making, and enhance customer experiences.

Expert: AI systems are becoming more capable due to advances in deep learning and access to large datasets. These models can now perform complex tasks.

Expert: At the same time, there are valid concerns about bias in AI models, as they often reflect the data they are trained on. Ensuring fairness is important.

Reporter: Governments and organizations are beginning to introduce regulations to guide the development and use of AI.

Expert: Another challenge is explainability. Many modern AI systems, especially deep neural networks, can be difficult to understand.

Reporter: Experts also highlight the importance of responsible AI development, including data privacy and security.

Expert: Looking ahead, collaboration between researchers, policymakers, and industry leaders will be essential.
"""
summary = summarize_dialogue(test_dialogue)
print("Summary: ", summary)
```
